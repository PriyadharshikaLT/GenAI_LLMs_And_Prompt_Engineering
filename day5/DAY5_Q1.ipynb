```python
from groq import Groq
from google.colab import userdata
```

```python
api_key = userdata.get("GROQ_API_KEY")

client = Groq(api_key=api_key)
```

```python
# Initial Prompt
prompt = "Explain Generative AI in 50 words"

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.5,
    max_tokens=100
)
generated_text1 = response.choices[0].message.content
generated_text1
```

```python
# Modified prompt
new_prompt = "Explain Machine Learning in 50 words"

response2 = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": new_prompt
        }
    ],
    temperature=0.5,
    max_tokens=100
)

generated_text2=response2.choices[0].message.content
generated_text2
```

# Working of the Groq LLM API

## Input
The user provides a prompt such as "Explain Generative AI in 50 words". This prompt is sent to the Groq API along with the selected language model.

## Processing
The Groq API receives the request and forwards it to the specified Large Language Model (LLM). The model analyzes the prompt, understands its context, and generates an appropriate response using its trained knowledge.

## Output
The API returns a structured response containing metadata and the generated text. The generated text is extracted from the response object using:

response.choices[0].message.content

The extracted text is then displayed to the user.
