```python
from groq import Groq
from google.colab import userdata
```

```python
api_key = userdata.get("GROQ_API_KEY")

client = Groq(api_key=api_key)
```

```python
email_type = input("Enter email type (Leave Request/Complaint/Meeting Request): ")

purpose = input("Briefly describe the purpose: ")

prompt = f"""
Write a professional {email_type} email.

Purpose:
{purpose}

Constraints:
1. Include a proper subject line.
2. Use a formal and professional tone.
3. Keep the email within 120 words.
4. Include a greeting.
5. Include a closing with regards.
6. Return only the email.
"""
```

```python
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.5,
    max_tokens=250
)
```

```python
email = response.choices[0].message.content

print("Generated Email\n")
email
```

```python
email_type = input("Choose email type (leave/complaint): ").lower()

details = input("Enter the details: ")

if email_type == "leave":

    prompt = f"""
Write a professional leave request email.

Reason:
{details}

Constraints:
- Include an appropriate subject.
- Formal tone.
- Greeting.
- Body.
- Closing.
- Maximum 120 words.
- Return only the email.
"""

elif email_type == "complaint":

    prompt = f"""
Write a professional complaint email.

Complaint:
{details}

Constraints:
- Include an appropriate subject.
- Formal tone.
- Greeting.
- Body.
- Closing.
- Maximum 120 words.
- Return only the email.
"""

else:
    prompt = f"""
Write a professional email.

Purpose:
{details}

Constraints:
- Include an appropriate subject.
- Formal tone.
- Greeting.
- Body.
- Closing.
- Maximum 120 words.
- Return only the email.
"""
```

```python
from IPython.display import Markdown, display
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.4,
    max_tokens=250
)

output = response.choices[0].message.content
display(Markdown(output))
```

# Working of the Professional Email Generator

## Step 1: User Input

The user selects an email type such as Leave Request or Complaint and provides the necessary details.

## Step 2: Prompt Creation

The application creates a prompt using the user's input. The prompt specifies that the email should:
- Have a proper subject line
- Use a formal tone
- Be limited to 120 words
- Include greeting and closing

## Step 3: API Call

The prompt is sent to the Groq LLM API using the `chat.completions.create()` method. The selected model processes the prompt and generates a professional email.

## Step 4: Output Processing

The generated email is extracted from:

`response.choices[0].message.content`

Only the email content is displayed in a clean and readable format.
